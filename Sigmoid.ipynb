```python
import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-10, 10, 400)
'''
linspace(start, stop, num) → Generates evenly spaced numbers
Breaking it down:
-10 → Starting value
10 → Ending value
400 → Total number of points to generate
So this line creates 400 equally spaced values between -10 and 10
'''
# Sigmoid Function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

y = sigmoid(x)

# Plot
plt.plot(x, y)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()

# Advantages
print("\nSigmoid Advantages:")
print("1. Output between 0 and 1 → good for binary classification.")
print("2. Smooth and differentiable everywhere.")

# Disadvantages
print("\nSigmoid Disadvantages:")
print("1. Causes vanishing gradient when x is very +ve or -ve.")
print("2. Not zero-centered → slows learning.")
print("3. Computation slightly expensive (exp function).")
```
