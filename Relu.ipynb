```python
import numpy as np
import matplotlib.pyplot as plt

# Input values
x = np.linspace(-10, 10, 400)
'''
linspace(start, stop, num) → Generates evenly spaced numbers
Breaking it down:
-10 → Starting value
10 → Ending value
400 → Total number of points to generate
So this line creates 400 equally spaced values between -10 and 10
'''

# ReLU Function
def relu(x):
    return np.maximum(0, x)

y = relu(x)

# Plot
plt.plot(x, y)
plt.title("ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid(True)
plt.show()

# Advantages
print("\nReLU Advantages:")
print("1. Very simple and fast to compute.")
print("2. Does not saturate for positive values.")
print("3. Helps deep networks learn faster (no vanishing gradient for x > 0).")

# Disadvantages
print("\nReLU Disadvantages:")
print("1. For negative x, output becomes 0 → neuron 'dies'.")
print("2. Not zero-centered.")
print("3. Gradient is 0 for x < 0 → learning stops there.")
```
