```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# -----------------------------------------
# Create simple dataset: y = 2x + 1
# -----------------------------------------
x = np.linspace(-1, 1, 100)
y = 2 * x + 1
```

```python
# -----------------------------------------
# A simple 1-layer model: y_pred = w*x + b
# -----------------------------------------
def train_model(lr, epochs=50):
    w, b = 0.0, 0.0      # initial weights
    loss_values = []     # for storing loss
    
    for epoch in range(epochs):
        y_pred = w * x + b
        loss = np.mean((y - y_pred)**2)       # MSE loss
        loss_values.append(loss)

        # gradients  
        dw = np.mean(2 * (y_pred - y) * x)
        db = np.mean(2 * (y_pred - y))

        # update weights
        w -= lr * dw
        b -= lr * db

    return loss_values
```

```python
# -----------------------------------------
# Try different learning rates
# -----------------------------------------
lrs = [0.0001, 0.01, 0.1, 1.0]
results = {}

for lr in lrs:
    results[lr] = train_model(lr)
```

```python
# -----------------------------------------
# Plot results
# -----------------------------------------
plt.figure(figsize=(10,6))

for lr in lrs:
    plt.plot(results[lr], label=f"LR = {lr}")

plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Learning Rate Experiment: Loss vs Epochs")
plt.legend()
plt.grid(True)
plt.show()
'''
What  Will Observe:
🔹 Learning Rate = 0.0001 (too small)
Loss decreases very slowly
Model learns extremely slowly
🔹 Learning Rate = 0.01 (good)
Smooth, fast convergence
Loss goes down steadily
🔹 Learning Rate = 0.1 (OK or risky)
Sometimes good
Sometimes oscillates
🔹 Learning Rate = 1.0 (too large)
Loss may explode
Model diverges instead of learning
Graph goes up and down or shoots up
Simple Explanation
Learning rate controls step size.
If steps are too small, training is slow.
If steps are too big, model jumps around and cannot learn.
A good learning rate gives smooth, fast decrease in loss.
'''
```
