```python
#simple neural network (NN)
#classifying mnist Digits (0-9)
#1. Import libraries

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten 
from tensorflow.keras.datasets import mnist
```

```python
#2 load the minst dasttaset(image of digits)
(x_train,y_train), (x_test,y_test)=mnist.load_data()
# images are 0-255 -> Normalize them to 0-1
x_train = x_train/255.0
x_test=x_test/255.0
```

```python
#3.Build a Simple Neural Network model
model=Sequential([
    #flatten converts 28x28 image -> 784 input vector
    Flatten(input_shape=(28,28)),
    #hidden layer: 128 neurans + Relu activation
    Dense(128, activation='relu'),
    #output layer :128 neurans (for digits 0-9)+ softmax
    Dense(10, activation='softmax'),
])
'''
Layers:
Flateen Layer
Converts 2D image - 1D vector
NN always takes 1D input
A fully connected hidden layer with 128 neurons
ReLU
'''
```

```python
#4.Model Compile
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
#5.Train the model
model.fit(x_train, y_train, epochs=5)
'''
Step 5-Train
NN adjust its weights using backpropagation.
'''
```

```python
#6.Evaluate the model
test_loss, test_acc=model.evaluate(x_test, y_test)
print("Test Accuracy:", test_acc)
'''
Step 6 -
We check accuracy on unseen test data
You will get the ariund 97-98% accuracy even with this tiny network.
'''
```
